```python
import pandas as pd

df = pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv", skiprows=4)
df.head()
```

```python
import matplotlib.pyplot as plt

plt.hist(df["2020"].dropna())   # use any year column
plt.title("Population Distribution (2020)")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
top10 = df.nlargest(10, "2020")

plt.bar(top10["Country Name"], top10["2020"])
plt.xticks(rotation=45)
plt.title("Top 10 Most Populated Countries (2020)")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv", skiprows=4)

# --- Histogram ---
plt.figure(figsize=(8,5))                       # create a new figure
plt.hist(df["2020"].dropna(), bins=30)          # draw the histogram
plt.title("Population Distribution (2020)")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.savefig("task1_population_histogram.png", bbox_inches="tight", dpi=300)
plt.show()
```

```python
top10 = df.nlargest(10, "2020")

plt.figure(figsize=(10,5))
plt.bar(top10["Country Name"], top10["2020"])
plt.title("Top 10 Countries by Population (2020)")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=45, ha="right")

plt.savefig("task1_top10_bar_2020.png", bbox_inches="tight", dpi=300)
plt.show()
```
